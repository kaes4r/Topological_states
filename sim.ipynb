```python
import matplotlib.pyplot as plt
import anyon_simulation as sim

Ec0 = sim.charging_energy(sim.A0)          # ~1.95 K for A0 = 0.1 um^2, d = 48 nm
B_grid, dVg_grid, dG = sim.simulate_grid(B_range=(8.2, 9.1), dVg_range=(-30, 0),
                                         kBT=0.002 * Ec0)

plt.pcolormesh(B_grid, dVg_grid, dG, cmap="RdBu_r", vmin=-1, vmax=1, shading="auto")
plt.xlabel("B (T)"); plt.ylabel(r"$\delta V_g$ (mV)")
plt.colorbar(label=r"$\langle\delta G\rangle / a_0$")
plt.show()
```
